```python
import re
import string

def wordopt(text):
    text = str(text).lower()
    text = re.sub(r'\[.*?\]', '', text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'<.*?>+', '', text)
    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub(r'\n', '', text)
    text = re.sub(r'\w*\d\w*', '', text)
    return text
```

```python
def output_label(n):
    if n == 0:
        return "Fake News"
    else:
        return "Real News"

import pandas as pd

def manual_testing(news):
    testing_news = {"text": [news]}
    new_df = pd.DataFrame(testing_news)

    # Apply preprocessing
    new_df["text"] = new_df["text"].apply(wordopt)

    # Vectorize
    new_x_test = vectorizer.transform(new_df["text"])

    # Predictions
    pred_LR  = LR.predict(new_x_test)[0]
    pred_DT  = DT.predict(new_x_test)[0]

   

    return {
        "Logistic Regression": output_label(pred_LR),
        "Decision Tree": output_label(pred_DT),
        
        
    }

    
    
```

```python
from sklearn.metrics import accuracy_score

scores = {
    "Logistic": accuracy_score(y_test, LR.predict(xv_test)),
    "DecisionTree": accuracy_score(y_test, DT.predict(xv_test)),
    
    
}

best_model_name = max(scores, key=scores.get)
best_model = {
    "Logistic": LR,
    "DecisionTree": DT,
    
   
}[best_model_name]

print("Best Model:", best_model_name)
```
